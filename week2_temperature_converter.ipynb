# Week 2 Assignment: Temperature Converter
Converts °C to °F, checks a threshold, loops over several values, and uses a function.

```python
# Function that converts Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):
    return celsius * 9 / 5 + 32
```

```python
# 1. Ask the user for a temperature in °C and convert it to °F
temp_c = float(input("Enter a temperature in °C: "))
temp_f = celsius_to_fahrenheit(temp_c)
print(f"{temp_c}°C is {temp_f:.1f}°F")
```

```python
# 2. Report whether the temperature is above or below a threshold
threshold_c = 30  # threshold in °C (change as needed)

if temp_c > threshold_c:
    print(f"{temp_c}°C is above the threshold of {threshold_c}°C")
elif temp_c < threshold_c:
    print(f"{temp_c}°C is below the threshold of {threshold_c}°C")
else:
    print(f"{temp_c}°C is exactly at the threshold of {threshold_c}°C")
```

```python
# 3. Use a loop to convert several temperatures
temperatures = [-10, 0, 15, 25, 37, 100]

for t in temperatures:
    print(f"{t}°C = {celsius_to_fahrenheit(t):.1f}°F")
```
